```python
# Part 5 – Matplotlib Visualization
```

```python
# 31.Create a bar chart showing the number of patients in each department.
```

```python
import matplotlib.pyplot as plt
```

```python
department_count = df["Department"].value_counts()
plt.figure(figsize=(10, 6))
plt.bar(department_count.index, department_count.values,color="olive",alpha=1)
plt.title("Number of Patients in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 32.Create a histogram showing the distribution of patient ages.
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["Age"].dropna(), bins=10, edgecolor="black",color="lightblue",alpha=0.7)
plt.title("Distribution of Patient Ages")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# 33.Create a line chart showing the number of patient admissions by month.
```

```python
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"],dayfirst=True,errors="coerce")
admissions = df.dropna(subset=["Admission_Date"]).copy()
monthly_admissions = (admissions.groupby(admissions["Admission_Date"].dt.to_period("M")).size())
plt.figure(figsize=(12, 6))
plt.plot(monthly_admissions.index.astype(str),monthly_admissions.values,marker="o")
plt.title("Number of Patient Admissions by Month")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
